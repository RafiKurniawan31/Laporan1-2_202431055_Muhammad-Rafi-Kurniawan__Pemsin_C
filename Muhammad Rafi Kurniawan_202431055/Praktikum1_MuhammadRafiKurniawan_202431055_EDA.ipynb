```python
#202431055_Muhammad Rafi Kurniawan
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#202431055_Muhammad Rafi Kurniawan
df = pd.read_csv('car_ad.csv', encoding='latin-1')
print(df)
print("Ukuran data (baris, kolom):", df.shape)
```

```python
#202431055_Muhammad Rafi Kurniawan
df.info()
```

```python
#202431055_Muhammad Rafi Kurniawan
print(df.describe())
```

```python
#202431055_Muhammad Rafi Kurniawan
print(df.isna().sum())
```

```python
#202431055_Muhammad Rafi Kurniawan
kolom_kategorik = ['car', 'body', 'engType', 'registration', 'model', 'drive']
for kolom in kolom_kategorik:
    print(f"\nKolom {kolom} ({df[kolom].nunique()} nilai unik):")
    print(df[kolom].unique())
```

```python
#202431055_Muhammad Rafi Kurniawan
df['engV'] = df['engV'].fillna(df['engV'].mean())

df['drive'] = df['drive'].fillna(df['drive'].mode()[0])
print(df.isna().sum())
```

```python
#202431055_Muhammad Rafi Kurniawan
print("Jumlah data duplikat:", df.duplicated().sum())
```

```python
#202431055_Muhammad Rafi Kurniawan
df = df.drop_duplicates()
df.info()
```

```python
#202431055_Muhammad Rafi Kurniawan
df['body'].value_counts().plot(kind='bar', color='steelblue')
plt.title('Jumlah Mobil per Tipe Bodi')
plt.xlabel('Tipe Bodi')
plt.ylabel('Jumlah Mobil')
plt.xticks(rotation=0)
plt.show()
```

```python
#202431055_Muhammad Rafi Kurniawan
df['drive'].value_counts().plot(kind='bar', color='darkorange')
plt.title('Jumlah Mobil per Jenis Penggerak')
plt.xlabel('Jenis Penggerak')
plt.ylabel('Jumlah Mobil')
plt.xticks(rotation=0)
plt.show()
```
